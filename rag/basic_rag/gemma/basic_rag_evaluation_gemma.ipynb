```python
import pandas as pd

# Kaydedilen değerlendirme dosyasını oku
df = pd.read_csv('rag_results_basic_gemma_evaluated.csv')

print("="*40)
print("     ORTALAMA METRİK SKORLARI")
print("="*40)

# Sadece metrik kolonlarını bul ve ortalamalarını yazdır
metrikler = ['faithfulness', 'answer_relevancy', 'context_recall', 'context_precision']

for metrik in metrikler:
    if metrik in df.columns:
        skor = df[metrik].mean()
        print(f"  - {metrik.upper():<20} : {skor:.4f}")
print("="*40)
```
